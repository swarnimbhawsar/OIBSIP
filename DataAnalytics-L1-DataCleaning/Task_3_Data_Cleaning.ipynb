```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# Load retail sales dataset

DATA_URL = "https://raw.githubusercontent.com/Seunfunmee/Retail-Sales-Data-Set/refs/heads/main/Retail%20Sales%20Data%20Set.csv"

df = pd.read_csv(DATA_URL)

print("Dataset loaded successfully!")
print("Original shape:", df.shape)

display(df.head())
```

```python
# Basic dataset inspection

print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())
```

```python
# Check missing values

missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": (missing_values / len(df) * 100).round(2)
})

display(missing_summary)
```

```python
# Check duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    display(df[df.duplicated()].head())
else:
    print("No duplicate rows found.")
```

```python
# Check current data types

dtype_summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values
})

display(dtype_summary)
```

```python
# Standardize column names

df.columns = [
    col.strip().lower().replace(" ", "_")
    for col in df.columns
]

print("Standardized column names:")
print(df.columns.tolist())
```

```python
# Remove unnecessary spaces from text columns

text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print("Text formatting cleaned successfully.")
```

```python
# Convert numerical columns to appropriate numeric data types

numeric_columns = [
    "age",
    "quantity",
    "price_per_unit",
    "total_amount"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Numerical columns converted successfully.")

display(df[numeric_columns].dtypes)
```

```python
# Convert date column to datetime format

df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

print("Date column data type:")
print(df["date"].dtype)

display(df[["date"]].head())
```

```python
# Check missing values after type conversion

print("Missing values before handling:")

display(
    df.isnull().sum().to_frame("Missing Values")
)
```

```python
# Handle missing numerical values using median

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

# Handle missing categorical values using mode

categorical_columns = df.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

# Remove rows where date is missing
df = df.dropna(subset=["date"])

print("Missing values handled.")

display(
    df.isnull().sum().to_frame("Remaining Missing Values")
)
```

```python
# Remove duplicate records

before_duplicates = len(df)

df = df.drop_duplicates()

after_duplicates = len(df)

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

```python
# Check for invalid values

print("Negative quantities:")
print((df["quantity"] < 0).sum())

print("\nNegative prices:")
print((df["price_per_unit"] < 0).sum())

print("\nNegative total amounts:")
print((df["total_amount"] < 0).sum())

print("\nInvalid ages:")
print(((df["age"] < 0) | (df["age"] > 120)).sum())
```

```python
# Remove invalid records

before_cleaning = len(df)

df = df[
    (df["quantity"] > 0) &
    (df["price_per_unit"] > 0) &
    (df["total_amount"] > 0) &
    (df["age"] >= 0) &
    (df["age"] <= 120)
]

after_cleaning = len(df)

print("Rows before validation:", before_cleaning)
print("Rows after validation:", after_cleaning)
print("Invalid rows removed:", before_cleaning - after_cleaning)
```

```python
# Detect outliers using the IQR method

def find_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return outliers, lower_bound, upper_bound


outlier_summary = []

for col in ["age", "quantity", "price_per_unit", "total_amount"]:

    outliers, lower, upper = find_outliers(df, col)

    outlier_summary.append({
        "Column": col,
        "Outliers": len(outliers),
        "Lower Bound": lower,
        "Upper Bound": upper
    })

outlier_summary = pd.DataFrame(outlier_summary)

display(outlier_summary)
```

```python
# Boxplots for numerical variables

numerical_for_plot = [
    "age",
    "quantity",
    "price_per_unit",
    "total_amount"
]

for col in numerical_for_plot:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[col])

    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)

    plt.tight_layout()
    plt.show()
```

```python
# Cap extreme outliers using IQR bounds

df_cleaned = df.copy()

for col in ["quantity", "price_per_unit", "total_amount"]:

    Q1 = df_cleaned[col].quantile(0.25)
    Q3 = df_cleaned[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_cleaned[col] = df_cleaned[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Extreme numerical values capped using IQR bounds.")
```

```python
# Validate the cleaned dataset

print("Missing values:")
display(df_cleaned.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df_cleaned.duplicated().sum())

print("\nData types:")
display(df_cleaned.dtypes.to_frame("Data Type"))

print("\nDataset shape:")
print(df_cleaned.shape)
```

```python
# Compare original and cleaned dataset

comparison = pd.DataFrame({
    "Metric": [
        "Number of Rows",
        "Number of Columns",
        "Missing Values",
        "Duplicate Rows"
    ],

    "Original": [
        len(pd.read_csv(DATA_URL)),
        len(pd.read_csv(DATA_URL).columns),
        pd.read_csv(DATA_URL).isnull().sum().sum(),
        pd.read_csv(DATA_URL).duplicated().sum()
    ],

    "Cleaned": [
        len(df_cleaned),
        len(df_cleaned.columns),
        df_cleaned.isnull().sum().sum(),
        df_cleaned.duplicated().sum()
    ]
})

display(comparison)
```

```python
# Final cleaned dataset

print("Final cleaned dataset:")
display(df_cleaned.head(10))

print("\nFinal dataset information:")
df_cleaned.info()
```

```python
# Save the cleaned dataset

output_file = "cleaned_retail_sales.csv"

df_cleaned.to_csv(
    output_file,
    index=False
)

print(f"Cleaned dataset saved as: {output_file}")
```

```python
# Final quality assurance checks

print("===== FINAL DATA QUALITY CHECK =====")

print("Rows:", len(df_cleaned))
print("Columns:", len(df_cleaned.columns))
print("Missing values:", df_cleaned.isnull().sum().sum())
print("Duplicate rows:", df_cleaned.duplicated().sum())

print("\nMinimum Age:", df_cleaned["age"].min())
print("Maximum Age:", df_cleaned["age"].max())

print("\nMinimum Quantity:", df_cleaned["quantity"].min())
print("Minimum Price:", df_cleaned["price_per_unit"].min())
print("Minimum Total Amount:", df_cleaned["total_amount"].min())

print("\nData cleaning completed successfully!")
```
