```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
# Load the retail sales dataset

DATA_URL = "https://raw.githubusercontent.com/Seunfunmee/Retail-Sales-Data-Set/refs/heads/main/Retail%20Sales%20Data%20Set.csv"

df = pd.read_csv(DATA_URL)

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)

display(df.head())
```

```python
# Basic information about the dataset

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Basic information about the dataset

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Check missing values and duplicate rows

print("Missing values in each column:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Standardise column names

df.columns = [
    col.strip().lower().replace(" ", "_")
    for col in df.columns
]

print("Updated column names:")
print(df.columns.tolist())

# Convert date column
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Make sure numerical columns are numeric
for col in ["age", "quantity", "price_per_unit", "total_amount"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Create month and quarter columns
df["month"] = df["date"].dt.to_period("M").astype(str)
df["quarter"] = df["date"].dt.to_period("Q").astype(str)

print("\nData preparation completed!")
display(df.head())
```

```python
# Descriptive statistics

numeric_columns = df.select_dtypes(include=np.number).columns

statistics = pd.DataFrame({
    "Mean": df[numeric_columns].mean(),
    "Median": df[numeric_columns].median(),
    "Mode": [
        df[col].mode().iloc[0]
        if not df[col].mode().empty
        else np.nan
        for col in numeric_columns
    ],
    "Standard Deviation": df[numeric_columns].std()
})

display(statistics)
```

```python
# Monthly revenue trend

monthly_sales = (
    df.groupby("month")["total_amount"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_sales,
    x="month",
    y="total_amount",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

display(monthly_sales)
```

### Observation

The monthly sales trend shows how revenue changes over time. Peaks indicate months with relatively higher sales, while lower points indicate weaker sales periods. These patterns can help a retailer plan inventory, promotions and marketing campaigns.

```python
# Quarterly revenue trend

quarterly_sales = (
    df.groupby("quarter")["total_amount"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=quarterly_sales,
    x="quarter",
    y="total_amount",
    marker="o"
)

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales Revenue")

plt.tight_layout()
plt.show()

display(quarterly_sales)
```

### Observation

The quarterly analysis provides a broader view of sales performance. Comparing quarters helps identify periods of stronger and weaker business activity and can support quarterly planning and resource allocation.

```python
# Create customer age groups

bins = [0, 17, 24, 34, 44, 54, 64, 100]
labels = [
    "Under 18",
    "18-24",
    "25-34",
    "35-44",
    "45-54",
    "55-64",
    "65+"
]

df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels
)

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="age_group",
    order=labels
)

plt.title("Customer Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers/Transactions")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Observation

The age-group distribution shows which customer age groups are most represented in the dataset. This information can help businesses understand their primary customer base and design appropriate marketing campaigns.

```python
# Gender distribution

gender_counts = df["gender"].value_counts()

plt.figure(figsize=(7, 5))

sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers/Transactions")

plt.tight_layout()
plt.show()

display(gender_counts)
```

### Observation

The gender distribution shows the relative representation of different genders in the retail dataset. Businesses can use this information as one of several factors when designing customer communication and promotional strategies.

```python
# Check available product-related columns

product_columns = [
    col for col in df.columns
    if "product" in col or "category" in col
]

print("Product-related columns:")
print(product_columns)
```

```python
# Top 10 product categories by revenue

top_categories = (
    df.groupby("product_category")["total_amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .sort_values()
)

plt.figure(figsize=(10, 6))

top_categories.plot(kind="barh")

plt.title("Top Product Categories by Revenue")
plt.xlabel("Total Revenue")
plt.ylabel("Product Category")

plt.tight_layout()
plt.show()

display(top_categories)
```

```python
# Revenue by product category

category_revenue = (
    df.groupby("product_category")["total_amount"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 6))

sns.barplot(
    x=category_revenue.values,
    y=category_revenue.index
)

plt.title("Revenue by Product Category")
plt.xlabel("Revenue")
plt.ylabel("Product Category")

plt.tight_layout()
plt.show()

display(category_revenue)
```

```python
# Correlation matrix

numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Numerical Variables")

plt.tight_layout()
plt.show()
```

```python
# Revenue by age group and gender

age_gender_sales = (
    df.groupby(
        ["age_group", "gender"],
        observed=True
    )["total_amount"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=age_gender_sales,
    x="age_group",
    y="total_amount",
    hue="gender"
)

plt.title("Revenue by Age Group and Gender")
plt.xlabel("Age Group")
plt.ylabel("Total Revenue")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Highest and lowest revenue months

highest_month = monthly_sales.loc[
    monthly_sales["total_amount"].idxmax()
]

lowest_month = monthly_sales.loc[
    monthly_sales["total_amount"].idxmin()
]

print("Highest revenue month:")
display(highest_month.to_frame().T)

print("\nLowest revenue month:")
display(lowest_month.to_frame().T)
```

```python
# Best-performing product category

best_category = category_revenue.idxmax()
best_category_revenue = category_revenue.max()

print("Best-performing category:", best_category)
print("Revenue:", best_category_revenue)
```

```python
# Overall business metrics

total_revenue = df["total_amount"].sum()
average_transaction = df["total_amount"].mean()
total_transactions = len(df)

print("===== BUSINESS SUMMARY =====")
print("Total Revenue:", total_revenue)
print("Average Transaction Value:", round(average_transaction, 2))
print("Total Transactions:", total_transactions)
print("Best Product Category:", best_category)
print("Highest Revenue Month:", highest_month["month"])
```
